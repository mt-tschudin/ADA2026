```python
import pandas as pd
import numpy as np
import os
```

```python
# Read file using relative paths
DATA_PATH = "./data"
DATA_FILE_NAME = "interactive_data.csv"
DATA_FILE_PATH = os.path.join(DATA_PATH, DATA_FILE_NAME)

df = pd.read_csv(DATA_FILE_PATH)
df.head()

```

```python
df_filtered = df.loc[~(df[['Intent','Gender','Age','Race']] == 'None selected').any(axis=1)]

gun_deaths = df_filtered['Deaths'].sum()
print("Total gun deaths: %d " %(gun_deaths))

```

```python
#Male suicide
df_suicides = df_filtered[df_filtered['Intent'] == 'Suicide']
suicide_deaths = df_suicides['Deaths'].sum()
df_male_suicide = df_filtered[(df_filtered['Gender'] == 'Male') & (df_filtered['Intent'] == 'Suicide')]
male_suicide_deaths = df_male_suicide['Deaths'].sum()

print("Male suicide deaths: %d" %(male_suicide_deaths))
print("Male suicide deaths percent: %d" %(male_suicide_deaths/suicide_deaths * 100))
```

```python
#Homicides
df_homicides = df_filtered[(df_filtered['Intent'] == 'Homicide')]
homicide_deaths = df_homicides['Deaths'].sum()

print("Homicide deaths: %d" %(homicide_deaths))
print("Homicide deaths percent: %d" %(homicide_deaths/gun_deaths * 100))
```

```python
#Young black male homicides

df_homicides_young_black_male = df_filtered[(df_filtered['Intent'] == 'Homicide') & (df_filtered['Age'] == "15 - 34") & (df_filtered['Gender'] == 'Male') & (df_filtered['Race'] == 'Black')]
df_homicides_young_male = df_filtered[(df_filtered['Intent'] == 'Homicide') & (df_filtered['Age'] == "15 - 34") & (df_filtered['Gender'] == 'Male')]

print("Young black male homicides: %d" %(df_homicides_young_black_male['Deaths'].sum()))
print("Young black male homicides percent: %d" %(df_homicides_young_black_male['Deaths'].sum() / df_homicides_young_male['Deaths'].sum() * 100))

```

```python
#Women homicides

df_homicides_woman = df_filtered[(df_filtered['Intent'] == 'Homicide') & (df_filtered['Gender'] == 'Female')]

print("Women homicides: %d" %(df_homicides_woman['Deaths'].sum()))
print("Women homicides percent: %d" %(df_homicides_woman['Deaths'].sum() / homicide_deaths * 100))

```
